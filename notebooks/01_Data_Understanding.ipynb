# AI-Powered Credit Risk Assessment & Smart Loan Recommendation System

## Business Problem

Banks receive thousands of loan applications every day. Approving loans for high-risk customers can lead to financial losses, while rejecting eligible customers can reduce business opportunities.

The objective of this project is to build a machine learning model that predicts whether a customer is eligible for a loan based on banking history and CIBIL information. Additionally, the project recommends an appropriate loan amount, repayment tenure, and provides business insights using exploratory data analysis.


## 1. Import Libraries

```python
# Data Manipulation
import pandas as pd
import numpy as np

# Data Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 10)

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully.")
```

## 2. Load Datasets

```python
import pandas as pd

# Load the three raw Excel files:
#  - Internal_Bank : the bank's own record of each customer's loans / trade lines
#  - External_Cibil: the CIBIL credit-bureau report (score, enquiries, delinquencies)
#  - Unseen        : a small hold-out set with no target column, used to test at the end
internal_df = pd.read_excel("../data/raw/Internal_Bank_Dataset.xlsx")
external_df = pd.read_excel("../data/raw/External_Cibil_Dataset.xlsx")
unseen_df = pd.read_excel("../data/raw/Unseen_Dataset.xlsx")

print("Datasets loaded successfully.")
```

## 3. Inspect Dataset Shapes

```python
# (rows, columns) of each file -- quick sanity check that they loaded fully.
# Internal + External both have 51,336 customers, so they can be joined 1-to-1.
print("Internal Dataset Shape :", internal_df.shape)
print("External Dataset Shape :", external_df.shape)
print("Unseen Dataset Shape   :", unseen_df.shape)
```

## 4. Preview Data

```python
# Look at the first 5 rows of each dataset to get a feel for what the columns hold
print("Internal Dataset")
display(internal_df.head())

print("External Dataset")
display(external_df.head())

print("Unseen Dataset")
display(unseen_df.head())
```

## 5. Data Types

```python
# .info() lists every column with its data type and non-null count -- lets me spot
# which columns are numbers vs text (object) before modelling.
print("Internal Dataset")
display(internal_df.info())

print("External Dataset")
display(external_df.info())

print("Unseen Dataset")
display(unseen_df.info())
```

## 6. Missing Values

```python
# Count blank (NaN) values in every column.
# Note: this dataset hides missing values as the number -99999 instead of NaN,
# so these come out as 0 here -- I actually clean up the -99999 sentinel in notebook 02.
missing_internal = internal_df.isnull().sum().sort_values(ascending=False)
missing_external = external_df.isnull().sum().sort_values(ascending=False)
missing_unseen = unseen_df.isnull().sum().sort_values(ascending=False)

print("Internal Dataset")
display(missing_internal[missing_internal > 0])

print("External Dataset")
display(missing_external[missing_external > 0])

print("Unseen Dataset")
display(missing_unseen[missing_unseen > 0])
```

## 7. Duplicate Records

```python
# Check for exact duplicate rows in each file -- all zero, so no de-duplication needed.
print("Internal Duplicate Rows :", internal_df.duplicated().sum())

print("External Duplicate Rows :", external_df.duplicated().sum())

print("Unseen Duplicate Rows :", unseen_df.duplicated().sum())
```

## 8. Merge Datasets

```python
# Join the bank data and the CIBIL data into ONE table, matched on the customer id
# (PROSPECTID). how="inner" keeps only customers that appear in BOTH files.
# Result: 51,336 rows x 87 columns -- this is the master dataset the project is built on.
merged_df = pd.merge(
    internal_df,
    external_df,
    on="PROSPECTID",
    how="inner"
)

print("Merged Dataset Shape :", merged_df.shape)

display(merged_df.head())
```

## 9. Statistical Summary

```python
# .describe() gives count / mean / std / min / max / quartiles for the numeric columns
# -- a quick statistical summary to spot odd ranges (e.g. income values that look wrong).
display(internal_df.describe())

display(external_df.describe())

display(unseen_df.describe())
```

## 10. Save the Processed Dataset


```python
# Save the merged master table to disk so the next notebook (EDA) can load it directly
# instead of re-reading and re-merging the raw Excel files every time.
merged_df.to_csv("../data/processed/merged_dataset.csv", index=False)

print("Merged dataset saved successfully.")
```
